# Peak Hours 

```python
%sql
CREATE OR REPLACE TABLE taxi23catalog.texi_schema.gold_peak_hours AS
SELECT
pickup_hour,
COUNT(*) AS total_trips,
SUM(total_amount) AS total_revenue
FROM taxi23catalog.texi_schema.silver_taxi_trips_enriched
GROUP BY pickup_hour
ORDER BY pickup_hour
     
```

#pickup vs dropoff time analysis

```python
%sql
CREATE OR REPLACE TABLE taxi23catalog.texi_schema.gold_pickup_dropoff_time AS
SELECT
pickup_hour,
dropoff_hour,
COUNT(*) AS total_trips
FROM taxi23catalog.texi_schema.silver_taxi_trips_enriched
GROUP BY pickup_hour, dropoff_hour
```

#Revenue by Time of Day (Very Powerful)

```python
%sql
CREATE OR REPLACE TABLE taxi23catalog.texi_schema.gold_time_of_day_revenue AS
SELECT
CASE
WHEN pickup_hour BETWEEN 6 AND 11 THEN 'Morning'
WHEN pickup_hour BETWEEN 12 AND 17 THEN 'Afternoon'
WHEN pickup_hour BETWEEN 18 AND 22 THEN 'Evening'
ELSE 'Night'
END AS time_of_day,
SUM(total_amount) AS total_revenue,
COUNT(*) AS total_trips
FROM taxi23catalog.texi_schema.silver_taxi_trips_enriched
GROUP BY time_of_day
```

# Revenue by each zone

```python
%sql
CREATE OR REPLACE TABLE taxi23catalog.texi_schema.gold_zone_revenue AS
SELECT
pickup_zone,
SUM(total_amount) AS total_revenue,
COUNT(*) AS total_trips
FROM taxi23catalog.texi_schema.silver_taxi_trips_enriched
GROUP BY pickup_zone
ORDER BY total_revenue DESC
```
